```python
import yake
import pandas as pd


def find_keywords (file_name):
    df = pd.read_csv (file_name)
    df ['key_words'] = None
    kw_extractor = yake.KeywordExtractor(n=1)
    for i in range (len(df)):
        keywords = kw_extractor.extract_keywords(df.loc[i, 'normalized_text'])
        lst = []
        for kw, score in keywords:
            lst.append ((kw, round(float(score),4)))
        df.loc[i, 'key_words'] = str(lst)
    return df


find_keywords ('cafe_reviews.csv')
```
