```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
import statsmodels.api as sm
import pandas as pd


def ii (file_name):
    data = pd.read_csv(file_name, index_col=0)
    data = data.drop(['Model', 'Engine Fuel Type', 'Transmission Type', 'Driven_Wheels', 'Number of Doors', 'Market Category', 'Vehicle Size', 'Vehicle Style'], axis=1)
    data['Engine HP'] = data['Engine HP'].fillna(data['Engine HP'] .median())
    data['Engine Cylinders'] = data['Engine Cylinders'].fillna(data['Engine Cylinders'] .median())
    data.info()

    features = data.drop('MSRP', axis=1)
    target = data['MSRP']

    features_train, features_test, target_train, target_test = train_test_split(features, target, test_size=0.25)

    model = LinearRegression()
    model.fit(features_train, target_train)
    result = model.predict(features_test)

    rmse_smart = root_mean_squared_error (target_test, result)

    result_slly = [target_train.mean()] * len(target_test)

    rmse_slly = root_mean_squared_error (target_test, result_slly)

    return rmse_smart, rmse_slly, rmse_slly - rmse_smart


ii ('data.csv')
```
